```r

# ============================================================
# TCC — MODELO 1
# Regra de Taylor com IBC-Br, sem câmbio
# Período: janeiro/2003 a julho/2026
# ============================================================

# 1. DATAS

Data <- seq(
  from = as.Date("2003-01-01"),
  to = as.Date("2026-07-01"),
  by = "month"
)


# 2. META SELIC
# Inserir 283 observações mensais

Selic <- c(
25.5,
26.5,
26.5,
26.5,
26.5,
26,
24.5,
22,
20,
19,
17.5,
16.5,
16.5,
16.5,
16.25,
16,
16,
16,
16,
16,
16.25,
16.75,
17.25,
17.75,
18.25,
18.75,
19.25,
19.5,
19.75,
19.75,
19.75,
19.75,
19.5,
19,
18.5,
18,
17.25,
17.25,
16.5,
15.75,
15.75,
15.25,
14.75,
14.25,
14.25,
13.75,
13.25,
13.25,
13,
13,
12.75,
12.5,
12.5,
12,
11.5,
11.5,
11.25,
11.25,
11.25,
11.25,
11.25,
11.25,
11.25,
11.75,
11.75,
12.25,
13,
13,
13.75,
13.75,
13.75,
13.75,
12.75,
12.75,
11.25,
10.25,
10.25,
9.25,
8.75,
8.75,
8.75,
8.75,
8.75,
8.75,
8.75,
8.75,
8.75,
9.5,
9.5,
10.25,
10.75,
10.75,
10.75,
10.75,
10.75,
10.75,
11.25,
11.25,
11.75,
12,
12,
12.25,
12.5,
12.5,
12,
11.5,
11.5,
11,
10.5,
10.5,
9.75,
9,
8.5,
8.5,
8,
7.5,
7.5,
7.25,
7.25,
7.25,
7.25,
7.25,
7.25,
7.5,
8,
8,
8.5,
9,
9,
9.5,
10,
10,
10.5,
10.75,
10.75,
11,
11,
11,
11,
11,
11,
11.25,
11.25,
11.75,
12.25,
12.25,
12.75,
13.25,
13.25,
13.75,
14.25,
14.25,
14.25,
14.25,
14.25,
14.25,
14.25,
14.25,
14.25,
14.25,
14.25,
14.25,
14.25,
14.25,
14.25,
14,
14,
13.75,
13,
12.25,
12.25,
11.25,
11.25,
10.25,
9.25,
9.25,
8.25,
7.5,
7.5,
7,
7,
6.75,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6.5,
6,
5.5,
5,
5,
4.5,
4.5,
4.25,
3.75,
3.75,
3,
2.25,
2.25,
2,
2,
2,
2,
2,
2,
2,
2.75,
2.75,
3.5,
4.25,
4.25,
5.25,
6.25,
7.75,
7.75,
9.25,
9.25,
10.75,
11.75,
11.75,
12.75,
13.25,
13.25,
13.75,
13.75,
13.75,
13.75,
13.75,
13.75,
13.75,
13.75,
13.75,
13.75,
13.75,
13.75,
13.25,
12.75,
12.75,
12.25,
11.75,
11.75,
11.25,
10.75,
10.75,
10.5,
10.5,
10.5,
10.5,
10.75,
10.75,
11.25,
12.25,
13.25,
13.25,
14.25,
14.25,
14.75,
15,
15,
15,
15,
15,
15,
15,
15,
15,
14.75,
14.5,
14.5,
14.25,
14.25
)


# 3. DESVIO DA INFLAÇÃO EM RELAÇÃO À META
# IPCA acumulado em 12 meses - meta anual
# Inserir 283 observações mensais

Desvio_Inflacao <- c(
5.97,
7.35,
8.07,
8.27,
8.74,
8.07,
6.93,
6.57,
6.64,
5.48,
2.52,
0.80,
2.21,
1.19,
0.39,
-0.24,
-0.35,
0.56,
1.31,
1.68,
1.20,
1.36,
1.74,
2.10,
2.91,
2.89,
3.04,
3.57,
3.55,
2.77,
2.07,
1.52,
1.54,
1.86,
1.72,
1.19,
1.20,
1.01,
0.82,
0.13,
-0.27,
-0.47,
-0.53,
-0.66,
-0.80,
-1.24,
-1.48,
-1.36,
-1.51,
-1.48,
-1.54,
-1.50,
-1.32,
-0.81,
-0.76,
-0.32,
-0.35,
-0.38,
-0.31,
-0.04,
0.06,
0.11,
0.23,
0.54,
1.08,
1.56,
1.87,
1.67,
1.75,
1.91,
1.89,
1.40,
1.34,
1.40,
1.11,
1.03,
0.70,
0.30,
0.00,
-0.14,
-0.16,
-0.33,
-0.28,
-0.19,
0.09,
0.33,
0.67,
0.76,
0.72,
0.34,
0.10,
-0.01,
0.20,
0.70,
1.13,
1.41,
1.49,
1.51,
1.80,
2.01,
2.05,
2.21,
2.37,
2.73,
2.81,
2.47,
2.14,
2.00,
1.72,
1.34,
0.74,
0.60,
0.49,
0.42,
0.70,
0.74,
0.78,
0.95,
1.03,
1.34,
1.65,
1.81,
2.09,
1.99,
2.00,
2.20,
1.77,
1.59,
1.36,
1.34,
1.27,
1.41,
1.09,
1.18,
1.65,
1.78,
1.87,
2.02,
2.00,
2.01,
2.25,
2.09,
2.06,
1.91,
2.64,
3.20,
3.63,
3.67,
3.97,
4.39,
5.06,
5.03,
4.99,
5.43,
5.98,
6.17,
6.21,
5.86,
4.89,
4.78,
4.82,
4.34,
4.24,
4.47,
3.98,
3.37,
2.49,
1.79,
0.85,
0.26,
0.07,
-0.42,
-0.90,
-1.50,
-1.79,
-2.04,
-1.96,
-1.80,
-1.70,
-1.55,
-1.64,
-1.66,
-1.82,
-1.74,
-1.64,
-0.11,
-0.02,
-0.31,
0.03,
0.06,
-0.45,
-0.75,
-0.47,
-0.36,
0.33,
0.69,
0.41,
-0.88,
-1.03,
-0.82,
-1.36,
-1.71,
-0.98,
0.06,
0.19,
0.01,
-0.70,
-1.60,
-2.12,
-1.87,
-1.69,
-1.56,
-0.86,
-0.08,
0.31,
0.52,
0.81,
1.45,
2.35,
3.01,
4.31,
4.60,
5.24,
5.93,
6.50,
6.92,
6.99,
6.31,
6.88,
7.04,
7.80,
8.63,
8.23,
8.39,
6.57,
5.23,
3.67,
2.97,
2.40,
2.29,
2.52,
2.35,
1.40,
0.93,
0.69,
-0.09,
0.74,
1.36,
1.94,
1.57,
1.43,
1.37,
1.51,
1.50,
0.93,
0.69,
0.93,
1.23,
1.50,
1.24,
1.42,
1.76,
1.87,
1.83,
1.56,
2.06,
2.48,
2.53,
2.32,
2.35,
2.23,
2.13,
2.17,
1.68,
1.46,
1.26,
1.44,
0.81,
1.14,
1.39,
1.72,
1.64,
1.44
)


# 4. HIATO DO PRODUTO — IBC-Br
# Inserir 283 observações mensais

Hiato_IBC <- c(
1.3052416,
2.6465875,
1.6722098,
0.5734678,
-0.8160983,
-1.0278685,
-2.6467698,
-2.2913734,
-0.5060664,
-1.0577816,
-1.1969169,
-1.8356364,
-0.4283889,
0.2241138,
0.9311014,
1.0372742,
0.5298117,
0.9616907,
0.9561527,
1.0278880,
1.6707133,
0.5770656,
1.2898679,
0.1475142,
0.3984788,
0.6323513,
0.3907875,
0.7171827,
0.2352365,
-0.7659425,
-1.1445119,
-0.7662338,
-1.3924393,
-1.7328627,
-1.3109356,
-0.9201655,
-0.5194916,
-0.8462987,
-1.3598945,
-1.1335616,
-0.2742705,
-1.8093102,
-0.3447475,
-0.7306792,
-0.8702564,
-0.2016185,
0.1415803,
1.0127832,
0.3170972,
0.1834383,
-0.7904544,
0.1869450,
0.6893889,
0.9530807,
1.3470609,
1.2020722,
1.6678184,
2.3265451,
1.9850438,
2.5900621,
2.4538481,
1.9651758,
0.9498585,
2.1006917,
2.7826991,
3.4451727,
3.2161314,
3.1725813,
2.8918246,
0.5918618,
-2.1318951,
-5.0374809,
-5.4709800,
-4.9939830,
-4.6615849,
-4.5242400,
-3.9199017,
-3.6237033,
-3.6796335,
-2.4334034,
-1.9489245,
-1.4994952,
-1.7191362,
-0.5882019,
0.5361907,
1.1402180,
1.8631473,
2.0192152,
0.8463240,
0.2524597,
0.5315760,
0.8356598,
1.4102525,
0.9313298,
1.7556420,
0.3958084,
1.3395263,
1.4643395,
1.1807318,
0.9971690,
0.8898462,
0.9816799,
0.8824743,
0.5741358,
0.2193177,
-0.0654115,
0.1991107,
-0.9557417,
-2.3634196,
-1.6039826,
-2.0228963,
-1.7473618,
-0.7799465,
-0.2328198,
-0.2235432,
-0.0317141,
-0.5862151,
0.1149401,
-0.1436093,
-0.6820259,
-0.1353152,
-1.1723532,
-0.2585376,
0.7845191,
0.8980505,
0.5047011,
0.8394081,
1.1475600,
1.6585400,
1.6362981,
1.9536077,
2.2852373,
2.0544346,
1.8524318,
1.8356743,
1.5312693,
1.3082977,
-0.7076199,
0.2141051,
1.2794105,
2.0196946,
1.3601261,
1.7564710,
2.3980337,
1.7443185,
1.3415444,
1.4081667,
0.3038347,
-0.1197500,
-0.8211108,
-1.7381027,
-0.9797185,
-1.4414523,
-0.8383252,
-1.6061693,
-1.5844435,
-1.8311776,
-1.7597364,
-2.2563672,
-2.1196450,
-2.1825482,
-1.8493854,
-2.0276194,
-2.1378694,
-1.8556898,
-2.2717440,
-2.1047121,
-2.2850270,
-1.6440116,
-0.4090506,
-0.2672209,
-0.3794618,
-0.5817860,
0.0780806,
0.1255109,
-0.1309578,
-0.1411601,
0.0635190,
0.2476349,
1.5715514,
1.0283042,
0.6129647,
0.6065100,
1.1587883,
-1.9757255,
0.8337719,
1.1315340,
1.9071469,
1.3128568,
1.6288974,
1.9123351,
2.1909511,
1.9496380,
1.3544626,
0.7204399,
0.6820281,
2.0890652,
1.5289371,
1.3165705,
1.5713695,
1.8970344,
3.0835944,
2.9056377,
2.5518765,
2.5444436,
2.6624477,
-4.1733188,
-12.9334475,
-10.3318992,
-6.3398212,
-3.9300361,
-2.1757833,
-0.3432921,
1.5676013,
1.6538928,
1.6783846,
1.6502476,
2.4376552,
-1.2665222,
-0.3865459,
0.1520734,
-0.3252018,
-0.8581112,
-0.4423841,
-0.7128944,
-0.3907805,
0.9867867,
1.0923417,
-1.1341192,
-0.2823530,
0.5611475,
0.2100004,
0.1551055,
-0.2370702,
1.4368170,
0.8674937,
0.6675293,
0.6931709,
-0.2746817,
-0.6526117,
-1.3847053,
1.2932281,
1.0464734,
1.6040360,
-0.1231288,
-0.4119058,
-0.1666912,
-0.8090401,
-1.0401254,
-0.9724090,
-0.4963925,
0.4904409,
-0.1775805,
-0.2382589,
-0.4591732,
-0.7631521,
-0.0900286,
0.5928752,
0.8280124,
0.7094564,
1.2785972,
1.2489439,
0.7946515,
-0.0263501,
0.3853477,
0.6534167,
1.7897059,
1.1066443,
0.3217666,
0.0636052,
-0.3380271,
-0.3130861,
-0.6213202,
-0.7900172,
-0.5094758,
-0.4949934,
-0.0681777,
0.3552499,
0.0180575,
0.1808615,
0.1841096,
-0.7878464,
-1.2757365
)


# 5. CONFERÊNCIA DOS VETORES

cat("Quantidade de datas:", length(Data), "\n")
cat("Quantidade de valores Selic:", length(Selic), "\n")
cat("Quantidade de valores do desvio:", length(Desvio_Inflacao), "\n")
cat("Quantidade de valores do hiato:", length(Hiato_IBC), "\n")

# Verifica se todos os vetores possuem o mesmo tamanho
stopifnot(
  length(Data) == length(Selic),
  length(Data) == length(Desvio_Inflacao),
  length(Data) == length(Hiato_IBC)
)


# 6. CONSTRUÇÃO DA BASE

base <- data.frame(
  Data = Data,
  Selic = Selic,
  Desvio_Inflacao = Desvio_Inflacao,
  Hiato_IBC = Hiato_IBC
)

# Conferência de valores ausentes
colSums(is.na(base))

head(base)
tail(base)
summary(base)


# 7. CONSTRUÇÃO DAS VARIÁVEIS DEFASADAS

# A Regra de Taylor utiliza as variáveis explicativas
# observadas no mês anterior.

n <- nrow(base)

dados_modelo <- data.frame(
  Data = base$Data[2:n],
  Selic = base$Selic[2:n],
  Selic_L1 = base$Selic[1:(n-1)],
  Desvio_Inflacao_L1 = base$Desvio_Inflacao[1:(n-1)],
  Hiato_IBC_L1 = base$Hiato_IBC[1:(n-1)]
)

# Número de observações efetivamente utilizadas
nrow(dados_modelo)


# 8. ESTIMAÇÃO DA REGRA DE TAYLOR POR MQO

# Forma linear auxiliar:
#
# i_t = alpha + rho*i_(t-1)
#       + gamma_pi*x_(t-1)
#       + gamma_y*y_(t-1) + erro
#
# Os coeficientes estruturais são recuperados posteriormente:
#
# phi_pi = gamma_pi / (1 - rho)
# phi_y  = gamma_y  / (1 - rho)

modelo1 <- stats::lm(
  Selic ~ Selic_L1 +
    Desvio_Inflacao_L1 +
    Hiato_IBC_L1,
  data = dados_modelo
)

# Resultados convencionais da regressão
summary(modelo1)


# 9. ERROS-PADRÃO ROBUSTOS — NEWEY-WEST

install.packages(c("sandwich", "lmtest"))

library(sandwich)
library(lmtest)

# Matriz de covariância robusta à heterocedasticidade
# e à autocorrelação serial

vcov_NW <- sandwich::NeweyWest(
  modelo1,
  lag = NULL,
  prewhite = FALSE,
  adjust = TRUE
)

# Resultados da regressão com erros-padrão robustos

resultado_NW <- lmtest::coeftest(
  modelo1,
  vcov. = vcov_NW
)

resultado_NW


# 10. RECUPERAÇÃO DOS PARÂMETROS ESTRUTURAIS

# Coeficientes estimados na forma auxiliar

b <- coef(modelo1)

alpha <- b["(Intercept)"]
rho <- b["Selic_L1"]
gamma_pi <- b["Desvio_Inflacao_L1"]
gamma_y <- b["Hiato_IBC_L1"]

# Recuperação dos coeficientes da Regra de Taylor

phi_pi <- gamma_pi / (1 - rho)
phi_y <- gamma_y / (1 - rho)

# Parâmetros estruturais

parametros <- c(
  alpha = alpha,
  rho = rho,
  phi_pi = phi_pi,
  phi_y = phi_y
)

parametros


# 11. ERROS-PADRÃO ROBUSTOS DOS PARÂMETROS ESTRUTURAIS

# Aplicação do método delta para calcular a matriz
# de covariância dos parâmetros transformados.

# Matriz Jacobiana da transformação

J <- matrix(
  0,
  nrow = 4,
  ncol = 4
)

colnames(J) <- names(b)
rownames(J) <- names(parametros)

J["alpha", "(Intercept)"] <- 1

J["rho", "Selic_L1"] <- 1

J["phi_pi", "Selic_L1"] <- gamma_pi / (1 - rho)^2
J["phi_pi", "Desvio_Inflacao_L1"] <- 1 / (1 - rho)

J["phi_y", "Selic_L1"] <- gamma_y / (1 - rho)^2
J["phi_y", "Hiato_IBC_L1"] <- 1 / (1 - rho)

# Matriz de covariância robusta transformada

vcov_estrutural <- J %*% vcov_NW %*% t(J)

# Erros-padrão robustos

erro_padrao <- sqrt(diag(vcov_estrutural))

# Estatísticas t e valores-p

estatistica_t <- parametros / erro_padrao

valor_p <- 2 * pt(
  abs(estatistica_t),
  df = df.residual(modelo1),
  lower.tail = FALSE
)

# Tabela final dos parâmetros estruturais

resultado_estrutural <- data.frame(
  Parametro = names(parametros),
  Estimativa = as.numeric(parametros),
  Erro_Padrao_NW = as.numeric(erro_padrao),
  Estatistica_t = as.numeric(estatistica_t),
  Valor_p = as.numeric(valor_p)
)

print(resultado_estrutural, digits = 5)


# 12. CRITÉRIOS DE INFORMAÇÃO

AIC(modelo1)
BIC(modelo1)


# 13. INFORMAÇÕES GERAIS DO MODELO

cat("\n========== MODELO 1 ==========\n")

cat("Período:",
    format(min(dados_modelo$Data), "%m/%Y"),
    "a",
    format(max(dados_modelo$Data), "%m/%Y"),
    "\n")

cat("Observações:", nobs(modelo1), "\n")
cat("R²:", summary(modelo1)$r.squared, "\n")
cat("R² ajustado:", summary(modelo1)$adj.r.squared, "\n")
cat("AIC:", AIC(modelo1), "\n")
cat("BIC:", BIC(modelo1), "\n")

cat("\nParâmetros estruturais:\n")
print(resultado_estrutural, digits = 5)
```

```r

# ============================================================
# TCC — MODELO 1
# JANELAS MÓVEIS (ROLLING WINDOWS)
# Janela: 84 observações mensais
# Deslocamento: 1 observação
# ============================================================


# 1. DEFINIÇÃO DOS PARÂMETROS

tamanho_janela <- 84

n_total <- nrow(dados_modelo)

n_janelas_esperado <- n_total - tamanho_janela + 1


# 2. VALIDAÇÕES INICIAIS

stopifnot(
  n_total == 282,
  tamanho_janela == 84,
  n_janelas_esperado == 199,
  all(complete.cases(dados_modelo)),
  !anyDuplicated(dados_modelo$Data)
)

cat("Observações disponíveis:", n_total, "\n")
cat("Tamanho de cada janela:", tamanho_janela, "\n")
cat("Número esperado de janelas:", n_janelas_esperado, "\n\n")


# 3. FUNÇÃO DE ESTIMAÇÃO DE UMA JANELA

estimar_janela <- function(inicio, dados) {

  fim <- inicio + tamanho_janela - 1

  base_janela <- dados[inicio:fim, ]

  # Estimação por MQO
  modelo <- stats::lm(
    Selic ~ Selic_L1 +
      Desvio_Inflacao_L1 +
      Hiato_IBC_L1,
    data = base_janela
  )

  # Verificação da estimação
  if (anyNA(coef(modelo)) || df.residual(modelo) <= 0) {
    stop(
      "Falha na estimação da janela iniciada na posição ",
      inicio
    )
  }

  # Matriz de covariância Newey-West
  vcov_nw <- sandwich::NeweyWest(
    modelo,
    lag = NULL,
    prewhite = FALSE,
    adjust = TRUE
  )

  # Resultados com erros-padrão robustos
  resultado_nw <- lmtest::coeftest(
    modelo,
    vcov. = vcov_nw
  )

  # Coeficientes auxiliares
  b <- coef(modelo)

  alpha <- unname(b["(Intercept)"])
  rho <- unname(b["Selic_L1"])
  gamma_pi <- unname(b["Desvio_Inflacao_L1"])
  gamma_y <- unname(b["Hiato_IBC_L1"])

  # Recuperação dos parâmetros estruturais
  if (abs(1 - rho) < 1e-8) {
    stop(
      "Coeficiente de persistência muito próximo de 1 na janela ",
      inicio
    )
  }

  phi_pi <- gamma_pi / (1 - rho)
  phi_y <- gamma_y / (1 - rho)

  # Método delta: matriz Jacobiana
  J <- matrix(
    0,
    nrow = 2,
    ncol = length(b),
    dimnames = list(
      c("phi_pi", "phi_y"),
      names(b)
    )
  )

  J["phi_pi", "Selic_L1"] <-
    gamma_pi / (1 - rho)^2

  J["phi_pi", "Desvio_Inflacao_L1"] <-
    1 / (1 - rho)

  J["phi_y", "Selic_L1"] <-
    gamma_y / (1 - rho)^2

  J["phi_y", "Hiato_IBC_L1"] <-
    1 / (1 - rho)

  # Covariância dos parâmetros estruturais
  vcov_estrutural <- J %*% vcov_nw %*% t(J)

  # Erros-padrão dos parâmetros estruturais
  ep_estrutural <- sqrt(diag(vcov_estrutural))

  parametros_estruturais <- c(
    phi_pi = phi_pi,
    phi_y = phi_y
  )

  # Testes t dos parâmetros estruturais
  t_estrutural <- parametros_estruturais / ep_estrutural

  p_estrutural <- 2 * stats::pt(
    abs(t_estrutural),
    df = df.residual(modelo),
    lower.tail = FALSE
  )

  # Data final da janela
  data_final <- base_janela$Data[nrow(base_janela)]

  # Organização dos resultados
  data.frame(

    Data = data_final,

    # Coeficientes auxiliares
    beta_selic = unname(b["Selic_L1"]),
    beta_inflacao = unname(b["Desvio_Inflacao_L1"]),
    beta_hiato = unname(b["Hiato_IBC_L1"]),

    # Erros-padrão Newey-West
    ep_beta_selic =
      resultado_nw["Selic_L1", "Std. Error"],

    ep_beta_inflacao =
      resultado_nw["Desvio_Inflacao_L1", "Std. Error"],

    ep_beta_hiato =
      resultado_nw["Hiato_IBC_L1", "Std. Error"],

    # Valores-p dos coeficientes auxiliares
    p_beta_selic =
      resultado_nw["Selic_L1", "Pr(>|t|)"],

    p_beta_inflacao =
      resultado_nw["Desvio_Inflacao_L1", "Pr(>|t|)"],

    p_beta_hiato =
      resultado_nw["Hiato_IBC_L1", "Pr(>|t|)"],

    # Parâmetros estruturais
    alpha = alpha,
    rho = rho,
    phi_pi = phi_pi,
    phi_y = phi_y,

    # Erros-padrão dos parâmetros estruturais
    ep_phi_pi = unname(ep_estrutural["phi_pi"]),
    ep_phi_y = unname(ep_estrutural["phi_y"]),

    # Valores-p dos parâmetros estruturais
    p_phi_pi = unname(p_estrutural["phi_pi"]),
    p_phi_y = unname(p_estrutural["phi_y"]),

    # Indicadores de ajuste
    R2 = summary(modelo)$r.squared,
    AIC = AIC(modelo),
    BIC = BIC(modelo)
  )
}


# 4. ESTIMAÇÃO DAS JANELAS MÓVEIS

resultados_janelas <- do.call(
  rbind,
  lapply(
    seq_len(n_janelas_esperado),
    estimar_janela,
    dados = dados_modelo
  )
)

rownames(resultados_janelas) <- NULL


# 5. VALIDAÇÃO DOS RESULTADOS

stopifnot(
  nrow(resultados_janelas) == 199,
  !anyNA(resultados_janelas),
  all(is.finite(as.matrix(resultados_janelas[, -1]))),

  resultados_janelas$Data[1] ==
    as.Date("2010-01-01"),

  resultados_janelas$Data[nrow(resultados_janelas)] ==
    as.Date("2026-07-01")
)

cat("\n========== VALIDAÇÃO DAS JANELAS ==========\n")

cat("Número de janelas estimadas:",
    nrow(resultados_janelas), "\n")

cat("Primeira janela termina em:",
    format(resultados_janelas$Data[1], "%m/%Y"), "\n")

cat("Última janela termina em:",
    format(resultados_janelas$Data[nrow(resultados_janelas)], "%m/%Y"),
    "\n")


# 6. FREQUÊNCIA DE SIGNIFICÂNCIA ESTATÍSTICA
# Coeficientes auxiliares
# Nível de significância: 5%

frequencia_significancia <- data.frame(

  Coeficiente = c(
    "Selic defasada",
    "Desvio da inflação",
    "Hiato do produto"
  ),

  Total_janelas = nrow(resultados_janelas),

  Janelas_significativas = c(
    sum(resultados_janelas$p_beta_selic < 0.05),
    sum(resultados_janelas$p_beta_inflacao < 0.05),
    sum(resultados_janelas$p_beta_hiato < 0.05)
  )
)

frequencia_significancia$Percentual <-
  100 *
  frequencia_significancia$Janelas_significativas /
  frequencia_significancia$Total_janelas

cat("\n========== FREQUÊNCIA DE SIGNIFICÂNCIA ==========\n")

print(frequencia_significancia, digits = 5)


# 7. RESUMO DOS PARÂMETROS ESTRUTURAIS

summary_parametros_estruturais <- data.frame(

  Parametro = c("rho", "phi_pi", "phi_y"),

  Media = c(
    mean(resultados_janelas$rho),
    mean(resultados_janelas$phi_pi),
    mean(resultados_janelas$phi_y)
  ),

  Minimo = c(
    min(resultados_janelas$rho),
    min(resultados_janelas$phi_pi),
    min(resultados_janelas$phi_y)
  ),

  Maximo = c(
    max(resultados_janelas$rho),
    max(resultados_janelas$phi_pi),
    max(resultados_janelas$phi_y)
  )
)

cat("\n========== RESUMO DOS PARÂMETROS ESTRUTURAIS ==========\n")

print(summary_parametros_estruturais, digits = 5)


# 8. GRÁFICOS DOS COEFICIENTES AUXILIARES

# Coeficiente da Selic defasada
plot(
  resultados_janelas$Data,
  resultados_janelas$beta_selic,
  type = "l",
  xlab = "Data",
  ylab = "Coeficiente estimado",
  main = "Evolução do coeficiente da Selic defasada"
)

abline(h = 0, lty = 2)


# Coeficiente do desvio da inflação
plot(
  resultados_janelas$Data,
  resultados_janelas$beta_inflacao,
  type = "l",
  xlab = "Data",
  ylab = "Coeficiente estimado",
  main = "Evolução do coeficiente do desvio da inflação"
)

abline(h = 0, lty = 2)


# Coeficiente do hiato do produto
plot(
  resultados_janelas$Data,
  resultados_janelas$beta_hiato,
  type = "l",
  xlab = "Data",
  ylab = "Coeficiente estimado",
  main = "Evolução do coeficiente do hiato do produto"
)

abline(h = 0, lty = 2)


# 9. CONFERÊNCIA DAS PRIMEIRAS E ÚLTIMAS JANELAS

cat("\n========== PRIMEIRAS JANELAS ==========\n")
print(head(resultados_janelas))

cat("\n========== ÚLTIMAS JANELAS ==========\n")
print(tail(resultados_janelas))


# 10. EXPORTAÇÃO DOS RESULTADOS

write.csv(
  resultados_janelas,
  "Modelo1_resultados_janelas_moveis.csv",
  row.names = FALSE,
  fileEncoding = "UTF-8"
)

write.csv(
  frequencia_significancia,
  "Modelo1_frequencia_significancia.csv",
  row.names = FALSE,
  fileEncoding = "UTF-8"
)
```

```r

# ============================================================
# 17. ESTIMAÇÃO DOS REGIMES — MODELO 1
# ============================================================

# 1. Separação dos regimes
indice_quebra <- as.integer(bp_otimo_m1$breakpoints)

stopifnot(
  length(indice_quebra) == 1,
  indice_quebra == 42,
  nrow(dados_modelo) == 282
)

dados_regime1 <- dados_modelo[1:indice_quebra, ]
dados_regime2 <- dados_modelo[(indice_quebra + 1):282, ]

# 2. Função de estimação e recuperação estrutural
estimar_regime <- function(dados, nome_regime) {

  modelo <- lm(
    Selic ~ Selic_L1 +
      Desvio_Inflacao_L1 +
      Hiato_IBC_L1,
    data = dados
  )

  V <- sandwich::NeweyWest(
    modelo,
    lag = NULL,
    prewhite = FALSE,
    adjust = TRUE
  )

  b <- unname(coef(modelo))

  alpha <- b[1]
  rho <- b[2]
  beta_pi <- b[3]
  beta_y <- b[4]

  denominador <- 1 - rho

  if (abs(denominador) < 1e-8) {
    stop("Rho muito próximo de 1 no regime: ", nome_regime)
  }

  parametros <- c(
    alpha = alpha,
    rho = rho,
    phi_pi = beta_pi / denominador,
    phi_y = beta_y / denominador
  )

  # Jacobiana para o método delta
  J <- matrix(0, nrow = 4, ncol = 4)

  J[1, 1] <- 1
  J[2, 2] <- 1

  J[3, 2] <- beta_pi / denominador^2
  J[3, 3] <- 1 / denominador

  J[4, 2] <- beta_y / denominador^2
  J[4, 4] <- 1 / denominador

  V_estrutural <- J %*% V %*% t(J)

  ep <- sqrt(diag(V_estrutural))

  estatistica_t <- parametros / ep

  valor_p <- 2 * pt(
    abs(estatistica_t),
    df = df.residual(modelo),
    lower.tail = FALSE
  )

  tabela <- data.frame(
    Regime = nome_regime,
    Parametro = names(parametros),
    Estimativa = as.numeric(parametros),
    Erro_padrao_NW = as.numeric(ep),
    Estatistica_t = as.numeric(estatistica_t),
    Valor_p = as.numeric(valor_p)
  )

  list(
    modelo = modelo,
    tabela = tabela,
    n = nobs(modelo),
    inicio = min(dados$Data),
    fim = max(dados$Data),
    R2 = summary(modelo)$r.squared
  )
}

# 3. Estimações
regime1_m1 <- estimar_regime(dados_regime1, "Regime 1")
regime2_m1 <- estimar_regime(dados_regime2, "Regime 2")

# 4. Tabela consolidada dos parâmetros estruturais
tabela_regimes_m1 <- rbind(
  regime1_m1$tabela,
  regime2_m1$tabela
)

cat("\n========== PARÂMETROS ESTRUTURAIS ==========\n")

print(
  transform(
    tabela_regimes_m1,
    Estimativa = round(Estimativa, 5),
    Erro_padrao_NW = round(Erro_padrao_NW, 5),
    Estatistica_t = round(Estatistica_t, 4),
    Valor_p = signif(Valor_p, 4)
  ),
  row.names = FALSE
)

# 5. Informações gerais
cat("\n========== INFORMAÇÕES DOS REGIMES ==========\n")

cat(
  "Regime 1:",
  format(regime1_m1$inicio, "%m/%Y"),
  "a",
  format(regime1_m1$fim, "%m/%Y"),
  "| N =", regime1_m1$n,
  "| R² =", round(regime1_m1$R2, 5),
  "\n"
)

cat(
  "Regime 2:",
  format(regime2_m1$inicio, "%m/%Y"),
  "a",
  format(regime2_m1$fim, "%m/%Y"),
  "| N =", regime2_m1$n,
  "| R² =", round(regime2_m1$R2, 5),
  "\n"
)

# 6. Coeficientes auxiliares com inferência Newey-West
cat("\n========== REGIME 1: COEFICIENTES AUXILIARES ==========\n")
print(lmtest::coeftest(
  regime1_m1$modelo,
  vcov. = sandwich::NeweyWest(
    regime1_m1$modelo,
    lag = NULL,
    prewhite = FALSE,
    adjust = TRUE
  )
))

cat("\n========== REGIME 2: COEFICIENTES AUXILIARES ==========\n")
print(lmtest::coeftest(
  regime2_m1$modelo,
  vcov. = sandwich::NeweyWest(
    regime2_m1$modelo,
    lag = NULL,
    prewhite = FALSE,
    adjust = TRUE
  )
))
```

```r

# ============================================================
# 18. TESTES DE CHOW — DATAS INSTITUCIONAIS — MODELO 1
# ============================================================

# Fórmula original
formula_m1 <- Selic ~ Selic_L1 +
  Desvio_Inflacao_L1 + Hiato_IBC_L1

# Datas definidas a partir das transições do BCB
datas_chow <- data.frame(
  Presidente = c(
    "Alexandre Tombini",
    "Ilan Goldfajn",
    "Roberto Campos Neto",
    "Gabriel Galipolo"
  ),
  Inicio = as.Date(c(
    "2011-01-01",
    "2016-06-01",
    "2019-03-01",
    "2025-01-01"
  ))
)

# Teste clássico de Chow para uma data fixa
teste_chow <- function(data_quebra, presidente) {

  antes <- dados_modelo[
    dados_modelo$Data < data_quebra, ,
    drop = FALSE
  ]

  depois <- dados_modelo[
    dados_modelo$Data >= data_quebra, ,
    drop = FALSE
  ]

  n1 <- nrow(antes)
  n2 <- nrow(depois)

  # Número de parâmetros, incluindo intercepto
  k <- length(coef(modelo1))

  # Verificação de graus de liberdade
  if (n1 <= k || n2 <= k) {
    stop("Observações insuficientes: ", presidente)
  }

  # Regressões irrestritas por subamostra
  modelo_antes <- lm(formula_m1, data = antes)
  modelo_depois <- lm(formula_m1, data = depois)

  # Soma dos quadrados dos resíduos
  rss_restrito <- sum(resid(modelo1)^2)

  rss_irrestrito <- sum(resid(modelo_antes)^2) +
    sum(resid(modelo_depois)^2)

  # Estatística F de Chow
  F_chow <- (
    (rss_restrito - rss_irrestrito) / k
  ) / (
    rss_irrestrito / (n1 + n2 - 2 * k)
  )

  valor_p <- pf(
    F_chow,
    df1 = k,
    df2 = n1 + n2 - 2 * k,
    lower.tail = FALSE
  )

  data.frame(
    Presidente = presidente,
    Data = format(data_quebra, "%m/%Y"),
    N_antes = n1,
    N_depois = n2,
    F = F_chow,
    Valor_p = valor_p,
    Significativo_5pct = valor_p < 0.05
  )
}

# Executar os quatro testes
resultados_chow_m1 <- do.call(
  rbind,
  lapply(seq_len(nrow(datas_chow)), function(i) {
    teste_chow(
      datas_chow$Inicio[i],
      datas_chow$Presidente[i]
    )
  })
)

# Validação
stopifnot(
  nrow(resultados_chow_m1) == 4,
  all(resultados_chow_m1$N_antes +
      resultados_chow_m1$N_depois == 282),
  all(is.finite(resultados_chow_m1$F)),
  all(is.finite(resultados_chow_m1$Valor_p))
)

cat("\n========== TESTES DE CHOW — MODELO 1 ==========\n")

print(
  transform(
    resultados_chow_m1,
    F = round(F, 4),
    Valor_p = signif(Valor_p, 5)
  ),
  row.names = FALSE
)
```
